# Step 1: Read your data into AMMICO

`ammico` reads in files from a directory. You can iterate through directories in a recursive manner and filter by extensions. Reading in these files creates a dictionary `image_dict`, with one entry per image file, containing the file path and filename. This dictionary is the main data structure that ammico operates on.

For reading in the files, the ammico function `find_files` is used, with optional keywords:

| input key | input type | possible input values |
| --------- | ---------- | --------------------- |
`path` | `str` | the directory containing the image files (defaults to the location set by environment variable `AMMICO_DATA_HOME`) |
| `pattern` | `str\|list` | the file extensions to consider (defaults to "png", "jpg", "jpeg", "gif", "webp", "avif", "tiff") |
| `recursive` | `bool` | include subdirectories recursively (defaults to `True`) |
| `limit` | `int` | maximum number of files to read (defaults to `20`, for all images set to `None` or `-1`) |
| `random_seed` | `int` | the random seed for shuffling the images; applies when only a few images are read and the selection should be preserved (defaults to `None`) |

SAIA limits the number of requests per minute, hour, day and month (see Step 2), so we recommend to start with a small number of images. To compare the models fairly, all models are run on exactly the same images.

````python
# Define your data path
data_path = "../../data/in"

# Find files and create the image dictionary
image_dict = ammico.find_files(
    path=data_path,
    limit=5,  # Limit the number of files to process (optional)
)
````

# Step 2: Connect to SAIA

To use SAIA, you need an API key. `ammico` connects to the model via three settings: the base URL, the API key and the model id. For SAIA, the base URL and the API key are the same for all models, only the model id changes. Setting `AMMICO_PRIVACY_ACK` to `"True"` acknowledges that the data is sent to an external service and silences the corresponding warning.

````python
import os
from getpass import getpass

if not os.environ.get("SAIA_API_KEY"):
    os.environ["SAIA_API_KEY"] = getpass("Enter your SAIA API key: ")

os.environ["AMMICO_API_BASE_URL"] = "https://chat-ai.academiccloud.de/v1"
os.environ["AMMICO_API_KEY"] = os.environ["SAIA_API_KEY"]
os.environ["AMMICO_PRIVACY_ACK"] = "True"  # acknowledge sending data externally
````

The number of requests you can send with your API key is limited per minute, hour, day and month. The limits depend on your key. The cell below sends one short request to a small text model to check that your key works, and prints your limits and how many requests you have left. This request counts towards your limits as well.

````python
from openai import OpenAI

client = OpenAI(
    base_url=os.environ["AMMICO_API_BASE_URL"], api_key=os.environ["AMMICO_API_KEY"]
)
response = client.chat.completions.with_raw_response.create(
    model="meta-llama-3.1-8b-instruct",
    messages=[{"role": "user", "content": "Hello"}],
    max_tokens=1,
)
for name, value in response.headers.items():
    if "ratelimit" in name:
        print(f"{name}: {value}")
````

# Step 3: Select the models to compare

Any model on SAIA that accepts images as input can be used. You can find the available models and their ids in the [SAIA documentation](https://docs.hpc.gwdg.de/services/ai-services/saia/index.html) (look for the capability "image"), and more information about each model in the [Chat AI model overview](https://docs.hpc.gwdg.de/services/ai-services/chat-ai/models/index.html). The models offered on SAIA change over time, so check the list if a model id is no longer found.

Add or remove model ids in the list below. The results of each model are saved to its own csv file in `results_dir`. Models for which a result file already exists are skipped, so you can add models to the list later and re-run the notebook without using up your requests again for the models you already ran. Set `overwrite = True` to re-run all models.

````python
models = [
    "gemma-4-31b-it",
    "qwen3.6-35b-a3b",
    "qwen3.8-27b",
    "glm-5.3-flash",
]

results_dir = "./saia_results"  # one csv file per model is written here
overwrite = False  # set to True to re-run models that already have a result file
````

`ammico` sends many requests at the same time to speed up the analysis. With the request limits of SAIA, this would quickly lead to errors, since requests that exceed the limit are rejected. The function below therefore creates models that send the requests one after the other (`max_concurrency=1`), with a pause between them, so that at most `requests_per_minute` requests are sent per minute. Set this to the per-minute limit printed in Step 2. Rejected requests are retried a few times (`max_retries`, 5 by default) before they are given up.

Some models "think" before they answer (reasoning models, for example the Qwen 3 models). The reasoning counts towards the token limit of the answer, so that the caption may come back empty. The function below asks the model to switch off the reasoning, by sending an additional setting with every request (`extra_body`); set `disable_reasoning = False` if you want to keep it. Models that do not support this setting ignore it.

````python
requests_per_minute = 10  # set to your per-minute limit from Step 2
disable_reasoning = True


def make_model(model_id):
    """An InferenceModel that respects the SAIA request limits and optionally switches off reasoning."""
    return ammico.InferenceModel(
        model_id=model_id,
        max_concurrency=1,  # one request at a time
        requests_per_minute=requests_per_minute,
        extra_body=(
            {"chat_template_kwargs": {"enable_thinking": False}}
            if disable_reasoning
            else None
        ),
    )
````

# Step 4: Run the analysis for all models

Now we define what we want to know about the images. You can choose between creating a caption (`"summary"`), answering your questions (`"questions"`) or both (`"summary_and_questions"`). With `is_concise_summary` and `is_concise_answer` you control whether the model answers briefly or in more detail. See the [image summary tutorial](https://github.com/ssciwr/AMMICO/blob/main/docs/tutorials/image_summary.md) for more details on the options.

Each image needs one request for the caption and one request per question. With the settings below and 5 images, this makes 15 requests per model.

````python
analysis_type = (
    "summary_and_questions"  # "summary", "questions" or "summary_and_questions"
)
list_of_questions = [
    "Who is in the picture?",
    "Is Trump in the picture, answer with only yes or no?",
]  # add or replace with your own questions
is_concise_summary = True
is_concise_answer = True
````

The cell below loops over all selected models. For each model, the analysis is run on a fresh copy of `image_dict`, so that the results of the different models do not overwrite each other. The results are converted into a dataframe and saved as `<results_dir>/<provider>__<model>.csv`; the `/` in the model id is replaced since it is not allowed in file names. The results of all models are also collected in the dictionary `results`.

Since the requests are sent one after the other, this takes some time: with a limit of 10 requests per minute, 15 requests per model take about one and a half minutes. If a request fails (for example because the model id is misspelled, the request limit is reached or the model does not accept images), a warning is printed and the answer stays empty or the image is skipped. If a model does not return any results at all, nothing is saved for this model, so that it is run again the next time you execute the cell; the loop then continues with the next model.

````python
import copy
import time
from pathlib import Path

import pandas as pd

Path(results_dir).mkdir(parents=True, exist_ok=True)


def result_file(model_id):
    """The csv file that holds the results of a model."""
    return Path(results_dir) / f"{model_id.replace('/', '__')}.csv"


results = {}
for model_id in models:
    outfile = result_file(model_id)
    if outfile.exists() and not overwrite:
        print(f"{model_id}: skipped, results already in {outfile}")
        continue

    print(f"{model_id}: running ...")
    start = time.time()
    try:
        model = make_model(model_id)
        model_dict = copy.deepcopy(
            image_dict
        )  # keep the results of each model separate
        image_summary_vqa = ammico.ImageSummaryDetector(
            summary_model=model, subdict=model_dict
        )
        model_dict = image_summary_vqa.analyse_images_from_dict(
            analysis_type=analysis_type,
            keys_batch_size=1,  # one image at a time, see the request limits above
            list_of_questions=list_of_questions,
            is_concise_summary=is_concise_summary,
            is_concise_answer=is_concise_answer,
        )
    except Exception as e:
        print(f"{model_id}: failed with {type(e).__name__}: {e}")
        continue

    # failed requests are skipped per image and leave empty results behind
    if not any(
        entry.get("caption") or entry.get("vqa") for entry in model_dict.values()
    ):
        print(
            f"{model_id}: no results returned, check the warnings above - nothing saved"
        )
        continue

    df = ammico.get_dataframe(model_dict)
    df.insert(0, "model", model_id)
    df.to_csv(outfile)
    results[model_id] = df
    print(f"{model_id}: done in {time.time() - start:.1f} s, saved to {outfile}")
````

# Step 5: Compare the results

The cell below reads the result files of all selected models back in. This way, the comparison also includes the models that were skipped above because they had been run before, for example in an earlier session.

````python
all_results = {
    model_id: pd.read_csv(result_file(model_id), index_col=0)
    for model_id in models
    if result_file(model_id).exists()
}
print(f"Results available for {len(all_results)} of {len(models)} models.")
````

To compare the captions, we put them into one table with one row per image and one column per model:

````python
column = "caption"  # use "vqa" to compare the answers to the questions instead

comparison = pd.DataFrame(
    {model_id: df.set_index("filename")[column] for model_id, df in all_results.items()}
)
pd.set_option("display.max_colwidth", None)
comparison
````

The comparison table can be saved to a csv file as well:

````python
comparison.to_csv(Path(results_dir) / f"comparison_{column}.csv")
````

For a closer look, we display each image together with the captions of all models:

````python
import textwrap

import matplotlib.pyplot as plt
from PIL import Image

for image_path, captions in comparison.iterrows():
    img = Image.open(image_path)
    plt.figure(figsize=(8, 6))
    plt.imshow(img)
    plt.axis("off")  # Hide axes
    plt.tight_layout()
    plt.show()
    for model_id, caption in captions.items():
        print(f"{model_id}:")
        print(textwrap.indent(textwrap.fill(str(caption), width=100), "    "))
    print()
````

If you want to keep all results in a single file instead, for example to analyze them further in another program, you can also combine the dataframes of all models into one long table. Each row then corresponds to one image analyzed by one model, and the column `model` tells you which model produced the result.

````python
combined = pd.concat(all_results.values(), ignore_index=True)
combined.to_csv(Path(results_dir) / "all_models.csv")
combined.head(5)
````
